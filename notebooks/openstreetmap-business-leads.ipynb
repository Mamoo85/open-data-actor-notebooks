## 1. Setup

```python
!pip install -q apify-client pandas

import os, getpass
try:
    from google.colab import userdata
    APIFY_TOKEN = userdata.get('APIFY_TOKEN')          # Colab: add it under the key icon (Secrets)
except Exception:
    APIFY_TOKEN = os.environ.get('APIFY_TOKEN') or getpass.getpass('Apify API token: ')
assert APIFY_TOKEN, 'An Apify API token is required (a free account is enough).'
print('Token loaded.')
```

## 2. Input
A deliberately small sample so the run is cheap and quick. See the [actor page](https://apify.com/transparent_meteorite/openstreetmap-business-leads) for every option.

```python
run_input = {   'locations': ['Detroit, Michigan'],
    'categories': ['cafes'],
    'onlyNamed': True,
    'onlyWithoutWebsite': True,
    'maxItems': 10}
```

## 3. Run the actor and read the dataset

```python
from apify_client import ApifyClient

client = ApifyClient(APIFY_TOKEN)
run = client.actor('transparent_meteorite/openstreetmap-business-leads').call(run_input=run_input)
# apify-client v2 returns a dict, v3 a Run object; handle both
get = (lambda k, a: run[k]) if isinstance(run, dict) else (lambda k, a: getattr(run, a))
items = list(client.dataset(get('defaultDatasetId', 'default_dataset_id')).iterate_items())
print('Run status:', get('status', 'status'), '| items:', len(items))
```

## 4. Tabulate

Phone and email columns are dropped before display; business name, category, address and website flags are shown.

```python
import pandas as pd
pd.set_option('display.max_colwidth', 70)
pd.set_option('display.width', 200)

df = pd.DataFrame(items)

# Privacy: remove contact / personal-data columns before anything is displayed
DROP = ["email", "phone", "author", "registeredagent", "contractor", "officer", "linkedin", "facebook", "instagram", "youtube", "tiktok", "socials", "twitter"]
KEEP = ["emailCount", "phoneCount", "emailProvider"]
df = df[[c for c in df.columns if c in KEEP or not any(p in c.lower() for p in DROP)]]
SHOW = [c for c in ["name", "categoryLabel", "website", "noWebsite", "fullAddress", "city", "completenessScore", "lastEditedAt", "osmUrl"] if c in df.columns]
df[SHOW]
```

## 5. A quick analysis

```python
df['completenessScore']=pd.to_numeric(df['completenessScore'],errors='coerce')
print(df['categoryLabel'].value_counts())
print('Median completeness:',df['completenessScore'].median())
df['noWebsite'].value_counts(dropna=False)
```
